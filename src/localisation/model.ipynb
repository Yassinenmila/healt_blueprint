```python

from ultralytics import YOLO

# Charger le modèle YOLOv8
model = YOLO("yolov8n.pt")

# Entraînement avec Data Augmentation
results = model.train(

    # Dataset
    data="../../data/dataset_yolo/data.yaml",

    # Paramètres d'entraînement
    epochs=50,
    imgsz=640,
    batch=8,

    # Data Augmentation
    degrees=5.0,
    translate=0.05,
    scale=0.10,

    fliplr=0.5,
    flipud=0.0,

    mosaic=0.5,

    hsv_h=0.0,
    hsv_s=0.0,
    hsv_v=0.10,

    # Sauvegarde
    project="runs/brain_tumor",
    name="yolov8n_augmentation"
)
```
