```python
import tensorflow as tf
from tensorflow.keras import layers

model = tf.keras.Sequential([
    layers.Input(shape=(None, None, 3)), # Accepte n'importe quelle taille d'origine
    layers.Resizing(224, 224),          # Étape 1 : Redimensionnement
    layers.Rescaling(1./255),           # Étape 3 : Normalisation
    layers.Conv2D(32, 3, activation='relu'),
])
```
