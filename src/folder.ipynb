```python

import random
import shutil
from pathlib import Path

# 1. Les dossiers
source = Path("../data/1")
destination = Path("../data/1_split")

random.seed(42)

# 2. Parcourir les 4 catégories
for categorie in ["Glioma", "Meningioma", "Pituitary", "No Tumor"]:

    images = []

    # 3. Récupérer les images de Train et Val
    for ensemble in ["Train", "Val"]:

        dossier = source / ensemble / categorie / "images"

        for image in dossier.iterdir():
            if image.is_file():
                images.append(image)

    # 4. Mélanger les images
    random.shuffle(images)

    # 5. Diviser les images
    total = len(images)

    train = images[:int(total * 0.70)]
    val = images[int(total * 0.70):int(total * 0.85)]
    test = images[int(total * 0.85):]

    # 6. Créer les trois ensembles
    ensembles = {
        "Train": train,
        "Val": val,
        "Test": test
    }

    # 7. Copier les images et leurs labels
    for nom, liste_images in ensembles.items():

        for image in liste_images:

            dossier_images = (
                destination / nom / categorie / "images"
            )
            dossier_images.mkdir(parents=True, exist_ok=True)

            shutil.copy2(image, dossier_images / image.name)

            # Copier le label correspondant, s'il existe
            dossier_labels = (
                source / "Train" / categorie / "labels"
            )
            label = dossier_labels / (image.stem + ".txt")

            if not label.exists():
                dossier_labels = (
                    source / "Val" / categorie / "labels"
                )
                label = dossier_labels / (image.stem + ".txt")

            if label.exists():
                cible = (
                    destination / nom / categorie / "labels"
                )
                cible.mkdir(parents=True, exist_ok=True)
                shutil.copy2(label, cible / label.name)

    print(categorie, ":", len(train), "Train,", len(val),
          "Val,", len(test), "Test")

print("Division terminée !")
```
